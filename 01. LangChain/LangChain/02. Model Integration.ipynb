### Streaming and Batch

#### Streaming
Most models can stream output while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an interior that yields output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
for chunk in model.stream("Write me a paragraph on Zero to One book"):
    print(chunk.text)
```

#### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
responses = model.batch(["Why does it rain?",
                       "How to make a rocket?",
                       "Can you sing?"
                       ])

for response in responses:
    print(response)


# We can also limit our batch size
# model.batch(["", "", "", "", "", ""], config={"max_concurrency": 3})
```
